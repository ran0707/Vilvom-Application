```python
from ultralytics import YOLO

# Load a pre-trained nano model
model = YOLO('/home/idrone2/Downloads/yolo11n-obb.pt')

# Train the model with your data.yaml file
model.train(data='/home/idrone2/Master_data/data.yaml', epochs=100, imgsz=640, batch=18, device=0, name='yolo11n-obb', exist_ok=True)
```

```python
import cv2
from ultralytics import YOLO

# Load the YOLOv8-OBB model
model = YOLO('runs/obb/yolo11n-obb/weights/best.pt')

# Set the model to evaluation mode
# model.eval()

# URL of the IP Webcam stream
url = 'http://10.181.127.119:8080/video'

# Open the video stream
cap = cv2.VideoCapture(url)

while True:
    # Capture frame-by-frame
    ret, frame = cap.read()
    if not ret:
        break

    # Perform inference on the frame
    results = model(frame)  # Pass the frame directly to the model

    # Parse results (YOLOv8-OBB returns oriented bounding boxes)
    for result in results:
        # Get the oriented bounding boxes, class IDs, and confidence scores
        obb_boxes = result.obb.xyxyxyxy.cpu().numpy()  # Move to CPU and convert to NumPy
        class_ids = result.obb.cls.cpu().numpy()       # Move to CPU and convert to NumPy
        confidences = result.obb.conf.cpu().numpy()    # Move to CPU and convert to NumPy

        # Draw oriented bounding boxes and labels
        for box, cls_id, conf in zip(obb_boxes, class_ids, confidences):
            # Convert the box to integer coordinates
            box = box.astype(int)

            # Draw the oriented bounding box (4 points)
            cv2.polylines(frame, [box], isClosed=True, color=(0, 255, 0), thickness=2)

            # Get the class label
            label = model.names[int(cls_id)]

            # Display the label and confidence score
            cv2.putText(frame, f'{label} {conf:.2f}', (box[0][0], box[0][1] - 10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0, 255, 0), 2)

    # Display the resulting frame using matplotlib
    import matplotlib.pyplot as plt

    # Convert BGR to RGB for matplotlib
    frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    plt.imshow(frame_rgb)
    plt.title('Pest Detection (YOLOv8-OBB)')
    plt.axis('off')
    plt.show()

    # Break after showing one frame (remove this if you want to show all frames)
    break

# Release the capture
cap.release()
```
